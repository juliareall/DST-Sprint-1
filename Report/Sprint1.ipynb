```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load datasets
train_file = list(Path.cwd().rglob("attainment8_train.csv"))[0]
test_file = list(Path.cwd().rglob("attainment8_test.csv"))[0]

train = pd.read_csv(
    train_file,
    na_values=["z", "c", "x"],
    dtype={"school_urn": "string", "time_period": "string"}
)

test = pd.read_csv(
    test_file,
    na_values=["z", "c", "x"],
    dtype={"school_urn": "string", "time_period": "string"}
)

# 1. Basic information
print("TRAINING DATA")
print("Shape:", train.shape)
display(train.head())
train.info()

print("\nTEST DATA")
print("Shape:", test.shape)
display(test.head())


# 2. Missing values
missing = pd.DataFrame({
    "Train missing": train.isna().sum(),
    "Train missing (%)": train.isna().mean() * 100,
    "Test missing": test.isna().sum(),
    "Test missing (%)": test.isna().mean() * 100
})

missing = missing.sort_values(
    "Train missing (%)", ascending=False
)

print("\nMISSING VALUES")
display(missing.round(2))

print("Total missing values (train):", train.isna().sum().sum())
print("Rows with missing values (train):", train.isna().any(axis=1).sum())
print("Columns with missing values (train):", train.isna().any(axis=0).sum())


# 3. Missing data visualisation
missing_percent = train.isna().mean() * 100
missing_percent = missing_percent[
    missing_percent > 0
].sort_values()

if not missing_percent.empty:
    plt.figure(figsize=(10, 12))
    missing_percent.plot(kind="barh")

    plt.xlabel("Missing values (%)")
    plt.ylabel("Variables")
    plt.title("Missing Data in Training Set")
    plt.tight_layout()
    plt.show()


# 4. Missing data by academic year
missing_by_year = (
    train.groupby("time_period")
    .agg(lambda x: x.isna().mean() * 100)
    .T
)

print("\nMISSING DATA BY YEAR")
display(missing_by_year.round(2))


# 5. Missing data by school type
key_columns = [
    "attainment8_average",
    "ks4_disadvantaged_pupil_percent",
    "sen_pupil_percent",
    "ks4_first_language_other_than_english_pupil_percent"
]

key_columns = [col for col in key_columns if col in train.columns]

missing_by_type = (
    train.groupby("establishment_type_group")[key_columns]
    .agg(lambda x: x.isna().mean() * 100)
)

print("\nMISSING DATA BY SCHOOL TYPE")
display(missing_by_type.round(2))


# 6. Summary statistics
print("\nSUMMARY STATISTICS")
display(train.describe().T.round(2))


# 7. Duplicate records
print("\nDUPLICATE CHECK")

print("Duplicate rows (train):", train.duplicated().sum())
print("Duplicate rows (test):", test.duplicated().sum())

print(
    "Duplicate school-year records (train):",
    train.duplicated(subset=["school_urn", "time_period"]).sum()
)

print(
    "Duplicate school-year records (test):",
    test.duplicated(subset=["school_urn", "time_period"]).sum()
)


# 8. Schools and academic years
print("\nSCHOOL INFORMATION")

print("Unique schools in training:", train["school_urn"].nunique())
print("Unique schools in testing:", test["school_urn"].nunique())

print("\nTraining records by year:")
print(train["time_period"].value_counts().sort_index())

print("\nTest records by year:")
print(test["time_period"].value_counts().sort_index())


# 9. School types
print("\nSCHOOL TYPES")
display(train["establishment_type_group"].value_counts())


# 10. Attainment 8 distribution (training data only)
print("\nATTAINMENT 8 STATISTICS")
print(train["attainment8_average"].describe())

plt.figure(figsize=(8, 5))

plt.hist(
    train["attainment8_average"].dropna(),
    bins=30,
    edgecolor="black"
)

plt.xlabel("Attainment 8 Score")
plt.ylabel("Number of School-Year Records")
plt.title("Distribution of Attainment 8 Scores")

plt.tight_layout()
plt.show()
```

```python

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_validate
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Select the explanatory variables
features = [
    "time_period",
    "la_name",
    "establishment_type_group",
    "pcon_name",
    "school_admission_type",
    "school_religious_character_detailed",
    "sex_policy",
    "min_age",
    "max_age",
    "allks_pupil_count",
    "endks4_pupil_count",
    "sen_pupil_count",
    "sen_pupil_percent",
    "sen_with_ehcp_pupil_count",
    "sen_with_ehcp_pupil_percent",
    "sen_no_ehcp_pupil_count",
    "sen_no_ehcp_pupil_percent",
    "pupil_count",
    "gcse_entries_average",
    "qual_entries_average",
    "sci_triple_entering_percent",
    "lan_multiple_entering_percent",
    "ks4_disadvantaged_pupil_count",
    "ks4_disadvantaged_pupil_percent",
    "ks4_not_known_disadvantaged_pupil_count",
    "ks4_not_known_disadvantaged_pupil_percent",
    "ks4_first_language_other_than_english_pupil_count",
    "ks4_first_language_other_than_english_pupil_percent",
    "ks4_non_mobile_pupil_count",
    "ks4_non_mobile_pupil_percent"
]

# Separate the predictors and target variable
X_train = train[features].copy()
y_train = train["attainment8_average"]

# Treat academic year as a category rather than a continuous number
X_train["time_period"] = X_train["time_period"].astype(str)

# Separate numerical and categorical predictors
categorical = X_train.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

numerical = [col for col in features if col not in categorical]

# Handle missing values and categorical variables
preprocessor = ColumnTransformer([
    (
        "numerical",
        SimpleImputer(strategy="median"),
        numerical
    ),
    (
        "categorical",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]),
        categorical
    )
])

# Build the Random Forest
rf = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        max_features=0.5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ))
])

# Five-fold cross-validation
cv = KFold(n_splits=5, shuffle=True, random_state=42)

scores = cross_validate(
    rf,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "MAE": "neg_mean_absolute_error",
        "RMSE": "neg_root_mean_squared_error",
        "R2": "r2"
    }
)

cv_results = pd.DataFrame({
    "MAE": -scores["test_MAE"],
    "RMSE": -scores["test_RMSE"],
    "R²": scores["test_R2"]
})

print("Cross-validation results")
display(cv_results.round(3))

print("\nMean cross-validation results")
print("MAE:", round(cv_results["MAE"].mean(), 3))
print("RMSE:", round(cv_results["RMSE"].mean(), 3))
print("R²:", round(cv_results["R²"].mean(), 3))

# Fit the model using the full training set
rf.fit(X_train, y_train)

train_predictions = rf.predict(X_train)

print("\nTraining performance")
print("MAE:", round(mean_absolute_error(y_train, train_predictions), 3))
print("RMSE:", round(np.sqrt(mean_squared_error(y_train, train_predictions)), 3))
print("R²:", round(r2_score(y_train, train_predictions), 3))

```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_validate, RandomizedSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Remove academic-related predictors
academic_features = [
    "gcse_entries_average",
    "qual_entries_average",
    "sci_triple_entering_percent",
    "lan_multiple_entering_percent"
]

social_features = [
    col for col in features
    if col not in academic_features
]

X_social = train[social_features].copy()
X_social["time_period"] = X_social["time_period"].astype(str)

y_social = train["attainment8_average"]

categorical_social = X_social.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

numerical_social = [
    col for col in social_features
    if col not in categorical_social
]

# Preprocessing
social_preprocessor = ColumnTransformer([
    (
        "numerical",
        SimpleImputer(strategy="median"),
        numerical_social
    ),
    (
        "categorical",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]),
        categorical_social
    )
])

# Initial model without academic predictors
social_rf = Pipeline([
    ("preprocessor", social_preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        max_features=0.5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=1
    ))
])

cv = KFold(n_splits=5, shuffle=True, random_state=42)

social_scores = cross_validate(
    social_rf,
    X_social,
    y_social,
    cv=cv,
    scoring={
        "MAE": "neg_mean_absolute_error",
        "RMSE": "neg_root_mean_squared_error",
        "R2": "r2"
    },
    n_jobs=-1
)

print("Random Forest without academic predictors")
print("Mean MAE:", round(-social_scores["test_MAE"].mean(), 3))
print("Mean RMSE:", round(-social_scores["test_RMSE"].mean(), 3))
print("Mean R²:", round(social_scores["test_R2"].mean(), 3))


# Hyperparameter tuning
parameters = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [10, 15, 20, None],
    "model__min_samples_leaf": [2, 5, 10],
    "model__max_features": [0.3, 0.5, 0.8]
}

social_search = RandomizedSearchCV(
    social_rf,
    param_distributions=parameters,
    n_iter=10,
    cv=cv,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    random_state=42,
    return_train_score=True
)

social_search.fit(X_social, y_social)

print("\nBest parameters:")
print(social_search.best_params_)

print("\nBest cross-validation RMSE:")
print(round(-social_search.best_score_, 3))

social_results = pd.DataFrame(social_search.cv_results_)

social_results["CV RMSE"] = -social_results["mean_test_score"]
social_results["Train RMSE"] = -social_results["mean_train_score"]

display(
    social_results[[
        "params",
        "Train RMSE",
        "CV RMSE"
    ]].sort_values("CV RMSE").round(3)
)

best_social_rf = social_search.best_estimator_
```

```python
# Fit the Random Forest using all training data
social_rf.fit(X_social, y_social)

model = social_rf.named_steps["model"]
preprocessor = social_rf.named_steps["preprocessor"]

importances = model.feature_importances_
encoded_features = preprocessor.get_feature_names_out()

# Combine encoded categories into their original variables
feature_importance = {}

for feature in social_features:
    indices = [
        i for i, name in enumerate(encoded_features)
        if name == "numerical__" + feature
        or name.startswith("categorical__" + feature + "_")
    ]

    feature_importance[feature] = importances[indices].sum()

importance = pd.Series(feature_importance).sort_values(ascending=False)

print("Feature importance")
display(importance.to_frame("Importance").round(4))

plt.figure(figsize=(10, 7))
importance.head(15).sort_values().plot(kind="barh")

plt.xlabel("Feature Importance")
plt.ylabel("Variable")
plt.title("Random Forest Feature Importance")

plt.tight_layout()
plt.show()
```

```python
# Compare Random Forest with and without geographical predictors

geographical_features = ["la_name", "pcon_name"]

no_geo_features = [
    col for col in social_features
    if col not in geographical_features
]

X_no_geo = X_social[no_geo_features].copy()

categorical_no_geo = X_no_geo.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

numerical_no_geo = [
    col for col in no_geo_features
    if col not in categorical_no_geo
]

# Preprocessing without geography
no_geo_preprocessor = ColumnTransformer([
    (
        "numerical",
        SimpleImputer(strategy="median"),
        numerical_no_geo
    ),
    (
        "categorical",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]),
        categorical_no_geo
    )
])

# Random Forest without geographical predictors
no_geo_rf = Pipeline([
    ("preprocessor", no_geo_preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        max_features=0.5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=1
    ))
])

# Use the same cross-validation setup
no_geo_scores = cross_validate(
    no_geo_rf,
    X_no_geo,
    y_social,
    cv=cv,
    scoring={
        "MAE": "neg_mean_absolute_error",
        "RMSE": "neg_root_mean_squared_error",
        "R2": "r2"
    },
    n_jobs=-1
)

# Compare results
geo_comparison = pd.DataFrame({
    "Model": ["With geography", "Without geography"],
    "CV MAE": [
        -social_scores["test_MAE"].mean(),
        -no_geo_scores["test_MAE"].mean()
    ],
    "CV RMSE": [
        -social_scores["test_RMSE"].mean(),
        -no_geo_scores["test_RMSE"].mean()
    ],
    "CV R²": [
        social_scores["test_R2"].mean(),
        no_geo_scores["test_R2"].mean()
    ]
})

display(geo_comparison.round(3))
```

```python
# Final Random Forest evaluation

# Fit the chosen model on the full training dataset
social_rf.fit(X_social, y_social)

# Prepare the test data using the same predictors
X_test = test[social_features].copy()
X_test["time_period"] = X_test["time_period"].astype(str)

y_test = test["attainment8_average"]

# Make predictions
test_predictions = social_rf.predict(X_test)

# Calculate performance metrics
test_mae = mean_absolute_error(y_test, test_predictions)
test_rmse = np.sqrt(mean_squared_error(y_test, test_predictions))
test_r2 = r2_score(y_test, test_predictions)

print("Final Random Forest test performance")
print("MAE:", round(test_mae, 3))
print("RMSE:", round(test_rmse, 3))
print("R²:", round(test_r2, 3))


# Compare cross-validation and test performance
final_results = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R²"],
    "Cross-validation": [
        -social_scores["test_MAE"].mean(),
        -social_scores["test_RMSE"].mean(),
        social_scores["test_R2"].mean()
    ],
    "Test": [test_mae, test_rmse, test_r2]
})

print("\nCross-validation vs test performance")
display(final_results.round(3))


# Actual vs predicted scores
plt.figure(figsize=(7, 6))

plt.scatter(
    y_test,
    test_predictions,
    alpha=0.3,
    s=15
)

limits = [
    min(y_test.min(), test_predictions.min()),
    max(y_test.max(), test_predictions.max())
]

plt.plot(limits, limits, "r--", label="Perfect prediction")

plt.xlabel("Actual Attainment 8")
plt.ylabel("Predicted Attainment 8")
plt.title("Actual vs Predicted Attainment 8")

plt.legend()
plt.tight_layout()
plt.show()


# Residual analysis
residuals = y_test.to_numpy() - test_predictions

plt.figure(figsize=(8, 5))

plt.scatter(
    test_predictions,
    residuals,
    alpha=0.3,
    s=15
)

plt.axhline(0, color="red", linestyle="--")

plt.xlabel("Predicted Attainment 8")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Random Forest Residuals")

plt.tight_layout()
plt.show()


# Distribution of prediction errors
plt.figure(figsize=(8, 5))

plt.hist(residuals, bins=30, edgecolor="black")

plt.xlabel("Residual")
plt.ylabel("Number of School-Year Records")
plt.title("Distribution of Prediction Errors")

plt.tight_layout()
plt.show()


# Investigate whether errors differ by school type
error_analysis = pd.DataFrame({
    "School type": test["establishment_type_group"],
    "Absolute error": np.abs(residuals)
})

errors_by_type = (
    error_analysis.groupby("School type")
    .agg(
        Records=("Absolute error", "size"),
        MAE=("Absolute error", "mean")
    )
    .sort_values("MAE", ascending=False)
)

print("\nPrediction errors by school type")
display(errors_by_type.round(3))
```

```python
from sklearn.model_selection import GroupKFold, cross_validate

# Keep observations from the same school together
group_cv = GroupKFold(n_splits=5)

group_scores = cross_validate(
    social_rf,
    X_social,
    y_social,
    groups=train["school_urn"],
    cv=group_cv,
    scoring={
        "MAE": "neg_mean_absolute_error",
        "RMSE": "neg_root_mean_squared_error",
        "R2": "r2"
    },
    n_jobs=1
)

# Compare validation methods
cv_comparison = pd.DataFrame({
    "Method": ["Ordinary K-Fold", "Grouped K-Fold"],
    "MAE": [
        -social_scores["test_MAE"].mean(),
        -group_scores["test_MAE"].mean()
    ],
    "RMSE": [
        -social_scores["test_RMSE"].mean(),
        -group_scores["test_RMSE"].mean()
    ],
    "R²": [
        social_scores["test_R2"].mean(),
        group_scores["test_R2"].mean()
    ]
})

display(cv_comparison.round(3))
```
